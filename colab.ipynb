## Run a new configuration
Loads a saved configuration and selects CUDA when available. Set the training horizon before starting a new run.


```python
import os
os.environ.setdefault('CUBLAS_WORKSPACE_CONFIG', ':4096:8')
import torch, json
from beyond_memorization.train import run
cfg = json.loads(Path('configs/mlp_main_ordered_s0.json').read_text())
cfg['device'] = 'cuda' if torch.cuda.is_available() else 'cpu'
out = Path('colab_runs/mlp_main_ordered_s0')
if (out / 'summary.json').exists():
    print('A completed run already exists:', out)
else:
    summary = run(cfg, out)
    print(json.dumps(summary, indent=2))
```
